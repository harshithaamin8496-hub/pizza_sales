## Visualization Basics

```python
#Bar Chart – Orders by Pizza Category
#Requirement:Display total orders in each pizza category.

orders = df['pizza_category'].value_counts()

x = orders.index
y = orders.values

plt.figure(figsize=(6,4))
plt.bar(x, y)

plt.title("Orders by Pizza Category")
plt.xlabel("Pizza Category")
plt.ylabel("Total Orders")

plt.show()
```

```python
#Pie Chart – Orders by Pizza Size
#Requirement: Show percentage contribution of each pizza size.

sizes = df['pizza_size'].value_counts()

labels = sizes.index
values = sizes.values

plt.figure(figsize=(6,6))
plt.pie(values,labels=labels,autopct='%1.2f%%')

plt.title("Orders by Pizza Size")

plt.show()
```

```python
# Histogram – Distribution of Unit Price
# Requirement: Analyze how unit prices are distributed.

plt.figure(figsize=(7,4))

plt.hist(df['unit_price'])

plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# Histogram – Distribution of Total Price
# Requirement: Analyze how total prices are distributed.

plt.figure(figsize=(7,4))

plt.hist(df['total_price'])

plt.title("Distribution of Total Price")
plt.xlabel("Total Price")
plt.ylabel("Frequency")

plt.show()
```

```python
#Scatter Plot – Unit Price vs Total Price
#Requirement: Check relationship between unit price and total price.

plt.figure(figsize=(7,4))

plt.scatter(df['unit_price'],df['total_price'])

plt.title("Unit Price vs Total Price")
plt.xlabel("Unit Price")
plt.ylabel("Total Price")

plt.show()
```

```python
#Scatter Plot – Quantity vs Total Price
#Requirement: Check relationship between quantity sold and total price.

plt.figure(figsize=(7,4))

plt.scatter(df['quantity'],df['total_price'])

plt.title("Quantity vs Total Price")
plt.xlabel("Quantity")
plt.ylabel("Total Price")

plt.show()
```

## Business KPIs

```python
#Calculate Total Revenue

total_revenue = df['total_price'].sum()
print("Total Revenue :", total_revenue)
```

```python
#Calculate Total Orders

total_orders = df['order_id'].nunique()
print("Total Orders :", total_orders)
```

```python
#Calculate Total Pizza Sold

total_pizza_sold = df['quantity'].sum()
print("Total Pizza Sold :", total_pizza_sold)
```

```python
#Calculate Average Order Value
#Formula:   Average Order Value = Total Revenue / Total Orders

total_revenue = df['total_price'].sum()
total_orders = df['order_id'].nunique()

avg_order_value = total_revenue / total_orders

print("Average Order Value :", avg_order_value)
```

```python
#Calculate Average Pizzas per Order
#Formula:   Average Pizzas per Order = Total Pizza Sold / Total Orders

total_pizza_sold = df['quantity'].sum()
total_orders = df['order_id'].nunique()

avg_pizza_per_order = total_pizza_sold / total_orders

print("Average Pizzas per Order :", avg_pizza_per_order)
```

```python
#Find Highest Ordered Size

highest_ordered_size = df.groupby('pizza_size')['order_id'].nunique().sort_values(ascending=False).head(1)
print(highest_ordered_size)
```
